```python
import os
import glob
import torch
from torch.utils.data import DataLoader
from src.data.sar_dataset import SARDataset

# 1. Find only actual image/chip files
files = [f for f in glob.glob("data/**/*", recursive=True) if os.path.isfile(f) and f.lower().endswith((".png", ".jpg", ".jpeg", ".npy", ".mat"))][:16]
assert len(files) > 0, "No valid data files found in data/"
labels = [i % 2 for i in range(len(files))]

# 2. Instantiate Dataset and DataLoader
ds = SARDataset(file_paths=files, labels=labels, target_size=128, rep_mode="ap")
dl = DataLoader(ds, batch_size=min(8, len(files)), shuffle=True)

# 3. Fetch one batch and inspect
batch, y = next(iter(dl))
print(f"Batch shape: {batch.shape}")
print(f"Batch dtype: {batch.dtype}")
print(f"Labels shape: {y.shape}")
print("DataLoader batch check passed successfully.")
```
